### 1. Introduction

Understanding the effects of disorder on quantum systems is central to the study of condensed matter physics. The Anderson model is a useful tool for exploring how randomness can induce localization in non-interacting electrons.

In this study, we apply similar methods to those employed in lesson 7 to study disorder in a 100 by 100 2D Anderson model. These include the study of the Density of States (DoS), IPR as a function of energy and Level-Spacing Distribution (LSD) with varying disorder strengths while considering the hopping paramter $t$ a constant. We expect to see a prevalence of extended states for lower disorder values, while high disorder values overpower electron hopping and in turn localise each state.

### 2. Fundamental Theory

#### Localization

Localization refers to the phenomenon where a particle's quantum wavefunction becomes confined due to disorder in the system. In a clean system, an electron's wavefunction can spread out over the entire lattice. It's extended, which allows for conduction.

When disorder is introduced, interference effects can cause the wavefunction to become localised in space, no longer contributing to conduction. This means the probability of finding the electron drops exponentially as you move away from the central point.

#### Two dimensional time-independent Schrodinger equation with a disorded potencial

Just like for the one-dimensional model studied briefly in lesson 7, a continuous two-dimensional model of a particle is considered, in a disorded potencial $V(x)$. By discreting the corresponding time-independent Schrodinger equation on a lattice with sufficently small spacing, an approximation of the 2D Anderson model can be obtained. The equation in question is given by:

$$
-\frac{\hbar^2}{2m_0} \left( \frac{\partial^2 \psi}{\partial x^2} + \frac{\partial^2 \psi}{\partial y^2} \right) + V(x, y)\psi = E \psi
$$

where $V(x,y)$ is some random potential and $E$ is the eigenenergy.

#### Discretization

Discretizing the equation for small but finite differences, $\Delta x = \Delta y = a$, allows the second order partial derivatives to be expressed as

$$
\frac{\partial^2 \psi}{\partial x^2} \approx \frac{\psi_{i+1,j} + \psi_{i-1,j} - 2\psi_{i,j}}{a^2} \quad \quad \frac{\partial^2 \psi}{\partial y^2} \approx \frac{\psi_{i,j+1} + \psi_{i,j-1} - 2\psi_{i,j}}{a^2}
$$

The $\frac{\partial^2\psi}{\partial x^2}$ and $\frac{\partial^2\psi}{\partial y^2}$ operatores can be expressed as a tridiagonal matrix

$$
H_x = \begin{bmatrix}
-2 & 1 & 0 & \cdots & 0 \\
1 & -2 & 1 & \cdots & 0 \\
0 & 1 & -2 & \cdots & 0 \\
\vdots & \vdots & \vdots & \ddots & 1 \\
0 & 0 & 0 & 1 & -2
\end{bmatrix}_{L \times L}
$$

$$
H_y = \begin{bmatrix}
-2 & 1 & 0 & \cdots & 0 \\
1 & -2 & 1 & \cdots & 0 \\
0 & 1 & -2 & \cdots & 0 \\
\vdots & \vdots & \vdots & \ddots & 1 \\
0 & 0 & 0 & 1 & -2
\end{bmatrix}_{L \times L}
$$
while the potencial operator is a diagonal matrix

$$
V = 
\begin{bmatrix}
V_{1,1} & 0 & \cdots & 0 \\
0 & V_{1,2} & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & V_{L,L}
\end{bmatrix}
$$

so the equation in matrix form is

$$
\begin{bmatrix}
4a_0 + V_{1,1} & -a_0 & 0 & \cdots & -a_0 & 0 & \cdots & 0 \\
-a_0 & 4a_0 + V_{1,2} & -a_0 & \cdots & 0 & -a_0 & \cdots & 0 \\
0 & -a_0 & \ddots & \ddots & \vdots & \vdots & \ddots & \vdots \\
\vdots & \vdots & \ddots & \ddots & -a_0 & 0 & \cdots & -a_0 \\
-a_0 & 0 & \cdots & -a_0 & 4a_0 + V_{2,1} & -a_0 & \cdots & 0 \\
0 & -a_0 & \cdots & 0 & -a_0 & 4a_0 + V_{2,2} & \ddots & \vdots \\
\vdots & \vdots & \ddots & \vdots & \vdots & \ddots & \ddots & -a_0 \\
0 & 0 & \cdots & -a_0 & 0 & \cdots & -a_0 & 4a_0 + V_{L,L}
\end{bmatrix}
\begin{bmatrix}
\Psi_{1,1} \\
\Psi_{1,2} \\
\vdots \\
\Psi_{1,L} \\
\Psi_{2,1} \\
\vdots \\
\Psi_{L,L}
\end{bmatrix}
= E
\begin{bmatrix}
\Psi_{1,1} \\
\Psi_{1,2} \\
\vdots \\
\Psi_{1,L} \\
\Psi_{2,1} \\
\vdots \\
\Psi_{L,L}
\end{bmatrix}
$$

where $a_0 = \frac{\hbar^2}{2m a^2}$ is a kinetic energy contant.

The previous discretization motivates the Anderson model, where the 2D Hamiltonian is given by:

$$
H = \sum_{i,j} \left( w_{i,j} |i,j\rangle \langle i,j| 
+ t_x |i,j\rangle \langle i+1,j| 
+ t_y |i,j\rangle \langle i,j+1| 
+ \text{h.c.} \right)
$$

where the state $|i,j\rangle$ is the occupation amplitude of the site $(i,j)$, with its on-site energy $w_\text{i,j} = V_{i,j}$, and $t \equiv a_0$ is the tunneling matrix element coupling neighbour sites.

Disorder is introduced into the system by allowing the on-site energies $w_\text{i,j}$ to random variables, uniformly distributed in the interval $[\frac{-W}{2}, \frac{W}{2}]$, with $W$ measuring the disorder strength.

#### Density of states

The DoS tells us how many quantum states are available at each energy level for particles like electrons, in the case of this study. For the 2D Anderson model with hopping $t = 1$ the disorder free bandwidth is about $[-4, 4]$, which can be broadend by a random on-site disorder $W$.

#### The Inverse Participation Ratio

In order to decide whether or not a state is localized it's often sufficent to consider the second moment of the probability density,

$$
\text{IPR} = P^{-1} = \sum_{i,j} |\psi_\text{i,j}|^4 \approx 
\begin{cases}
1/N  \quad \quad(extended)\\
Const. \quad (localized)
\end{cases}
$$

where $N$ is the number of lattice sites. 

This is the inverse participation number. It's a measure of the portion of the space where the amplitude of the wavefunction differs from zero. IPR vs energy plots reveal critical energies where the states transition from extended to localized.

#### Level-Spacing Distribution

As the name implies, the Level-Spacing Distribution shows you the distribution of level spacings, which are the differences between consecutive eigenenergies.

For a 2D system of size $L \times L$, there are $N = L^2$ eigenenergies. The level-spacing distributions statistics can distinguish between two distributions. 
For low disorder values ($W \lesssim 2$) the LSD is expected to resemble a Wigner-Dyson distribution, which is associated with a largely still extended system. For higher disorder values ($W \gtrsim 10$) instead resembles a Poisson distribution, associated with a largely localised system.
In between is a transition period where the LSD changes its behaviour, where there is a mix of localized and extended states.

### 3. Code


```
ModelAnalysis2D[nx_, ny_, W_] := Module[
    {
        w, hamiltonian, eigenValues, eigenVectors, IPR, levelSpacings,
        posToIndex, indexToPos, totalSites = nx*ny,
        binWidth = 0.1, bins, binCenters, dosData
    },

    (* Coordinate ↔ index helpers *)
    posToIndex[{x_, y_}] := (y - 1)*nx + x;
    indexToPos[index_] := {Mod[index - 1, nx] + 1, Quotient[index - 1, nx] + 1};

    (* Disorder potential *)
    w = RandomReal[{-W/2, W/2}, totalSites];

    (* Full dense Hamiltonian *)
    hamiltonian = ConstantArray[0., {totalSites, totalSites}];
    Do[
        hamiltonian[[i, i]] = w[[i]],
        {i, 1, totalSites}
    ];

    (* Nearest-neighbor hopping *)
    Do[
        {x, y} = indexToPos[i];
        (* Right neighbour hoping *)
        If[x < nx,
            j = posToIndex[{x + 1, y}];
            hamiltonian[[i, j]] = -1;   (* t = 1 implied by -1 *)
            hamiltonian[[j, i]] = -1;
        ];
        (* Left neighbour hoping *)
        If[y < ny,
            j = posToIndex[{x, y + 1}];
            hamiltonian[[i, j]] = -1;   (* t = 1 implied by -1 *)
            hamiltonian[[j, i]] = -1;
        ];
        ,
        {i, 1, totalSites}
    ];

    (* Full eigensystem (be cautious for large totalSites) *)
    {eigenValues, eigenVectors} = Eigensystem[hamiltonian];

    (* IPR (vectors are orthonormal already) *)
    IPR = Total[#^4] & /@ eigenVectors;

    (* Level spacings *)
    levelSpacings = Differences[Sort[eigenValues]];

    (* Density of States (scatter style) *)
    bins = BinCounts[eigenValues, {Min[eigenValues], Max[eigenValues], binWidth}];
    binCenters = Range[
        Min[eigenValues] + binWidth/2,
        Max[eigenValues] - binWidth/2,
        binWidth
    ];
    dosData = Transpose[{binCenters, bins}];

    (* Return output plots *)
    <|
        "DOS" -> ListPlot[
            dosData,
            PlotStyle -> PointSize[Medium],
            PlotLabel -> StringForm["Density of States (W = `1`)", W],
            AxesLabel -> {"Energy", "Density"},
            ImageSize -> Medium
        ],

        "IPR" -> ListPlot[
            Transpose[{eigenValues, IPR}],
            PlotLabel -> StringForm["IPR vs Energy (W = `1`)", W],
            AxesLabel -> {"Energy", "IPR"},
            PlotRange -> All,
            ImageSize -> Medium
        ],

        "LevelSpacing" -> Histogram[
            levelSpacings,
            Automatic,
            "PDF",
            ChartStyle -> RGBColor[0.3, 0.6, 0.9],
            PlotLabel -> StringForm["Level Spacing Distribution (W = `1`)", W],
            AxesLabel -> {"Level Spacing", "Frequency"},
            ImageSize -> Medium
        ]
    |>
]


(* Parameters *)
nx = 100; (* Number of sites in x-direction *)
ny = 100; (* Number of sites in y-direction *)
t = 1; (* hopping paramter, t_x = t_y *)
```

```
(* No disorder -> W = 0 *)
W = 0;

analysis2D = ModelAnalysis2D[nx, ny, W];

GraphicsRow[{
    analysis2D["DOS"],
    analysis2D["IPR"]
}, ImageSize -> Large]

analysis2D["LevelSpacing"]
```

### 4. Discussion

#### Density of States

For weaker disorder values, the electron wavefunctions are still largerly extended across the lattice. Electrons are able to move freely and the eigenstates resemble bloch-like states perturbed by randomness. The DoS has a distinct shape, resembling a logarithmic Van Hove singularity at $E = 0$. Since the each state is still largely extended, this allows for their wavefunctions to overlap , creating quantum interference, which in turn causes level repulsion.

This is coroborated by the DoS plots for disorder values $W = [0, 0.01, 0.15, 0.5, 1, 2]$

As disorder values increase, as seen for the cases $W = [5, 10, 16.5]$, the main curve begin to flatten, while leaving behind 'tails' at the edges of the curve. These are called Lifshitz tails and are associated with localization. They indicate rare 'events', or regions where the random potencial is unusual, which can trap a state.

At very high disorder strengths the random on-site potencial dominates the system. There is no wavefunction overlap, which means no repulsion, making hopping irrelevant. Each eigenstate becomes localized on a single site and the system now behaves like a collection of independent sites with their own potencial, mimicking a lattice composed of atoms.

As seen on the the cases $W = [25,50]$ the DoS no longer resembles a curve, but instead a scatter plot and the Lifshitz tails are getting progressively smaller, implying a smaller change of unusual potencials in the system.

For $W = 100$, while there are still small Lifshitz tails, the system is now largely localized. This results in each state converging to a discrete density on the plot, and the formation of horizontal lines on the DoS. For $W = 1000$ there are no longer any tails and most states are now fully localized.

#### IPR as a function of energy

At low disorder values, the IPR is flat across the spectrum, implying most states are extended. This is seen in $W = 0$. As disorder is introduced into the system, states near the band egdes begin to localize earlier than lower energy states near the center, resulting in the spikes seen for $W = [0.15, 0.5, 1, 2, 5]$. This is because, near the band edges there are fewer nearby states, which means wavefunctions have less room to spread. This results in reduced quantum interference, which promotes localization.

By increasing the disorder, the IPR becomes strongly energy dependent, as edges states localize faster than the lower energy states, forming a U-shape. This is seen in $W = 10$. Increasing disorder even further, the LSD loses its U-shape (as seen for $W = [16.5, 25]$) as all states of the spectrum become localized. For sufficiently high disorder strengths, the curve flattens and all states are localized, as seen in $W = [50, 100, 1000]$.

#### Level-Spacing Distribution

As expected for lower disorder values, the LSD follows a Wigner-Dyson distribution, where small spacings are uncommon. This means there is repulsion which implies the existance of extended states. This behaviour can be observed until $W = 2$. 

At $W = 10$ the LSD takes on a Poisson distribution, where the most common level spacing the shortest. There is no longer any repulsion. This is due to the fact that the now localized eigenstates don't overlaps and are therefore statistically independent.

In between $W = 2$ and $W = 10$ we have a transition period from one distribution to another, as we can see by the increasing, but still relatively small, shortest level spacing.

### 5. Conclusion

The obtained results back up what was previously known about the 1D Anderson model and increase our understanding of condensed matter. As expected, lower disorder allows the wavefunctions to overlap, allowing for electron hopping and subsequent conduction, indicative of a conductor, while higher disorder restricts wavefunctions, localizing them and restricting hopping. The 2D model now behaves like an atom based lattice.
